# Quick access: puntos con mayor error (2HDMC vs Colab)
Notebook corto para filtrar el CSV final y ver solo parámetros útiles + errores clave.


```python
from pathlib import Path
import numpy as np
import pandas as pd

CANDIDATES = [
    Path('out/refactor_colab_compare/real_run/colab_vs_2hdmc_merged_comparison.csv'),
    Path('colab_vs_2hdmc_merged_comparison.csv'),
]
CSV_PATH = next((p for p in CANDIDATES if p.exists()), None)
if CSV_PATH is None:
    raise FileNotFoundError('No encontré colab_vs_2hdmc_merged_comparison.csv en rutas esperadas.')

df = pd.read_csv(CSV_PATH)
if 'triple_ok' in df.columns:
    df = df[df['triple_ok'] == True].copy()

print('CSV:', CSV_PATH)
print('rows (triple_ok):', len(df))
```

```python
# Parámetros solicitados + aliases con nombres amigables
base = pd.DataFrame({
    'point_id': df['point_id'] if 'point_id' in df.columns else np.nan,
    'm_phi': df['mH'] if 'mH' in df.columns else df.get('mh', np.nan),
    'lambda_1': df['lambda1'] if 'lambda1' in df.columns else np.nan,
    'lambda_2': df['lam2_g'] if 'lam2_g' in df.columns else np.nan,
    'm_12_sq': df['m12_2_g'] if 'm12_2_g' in df.columns else df.get('m12sq_analytic', np.nan),
    'tan_beta': df['tan_beta'] if 'tan_beta' in df.columns else np.nan,
    'lambda_6': df['lambda6'] if 'lambda6' in df.columns else np.nan,
    'lambda_7': df['lambda7'] if 'lambda7' in df.columns else np.nan,
    'm_A': df['mA'] if 'mA' in df.columns else np.nan,
    'm_Hp': df['mHp'] if 'mHp' in df.columns else np.nan,
    'sin_ba': df['sba'] if 'sba' in df.columns else np.nan,
})

# Errores clave de widths
for c in ['rel_err_width_bb','rel_err_width_cc','rel_err_width_tautau','rel_err_width_gg','rel_err_width_gaga','rel_err_width_Zga','rel_err_width_total']:
    base[c] = df[c] if c in df.columns else np.nan

# Métricas para priorizar re-ejecución
err_cols = [c for c in base.columns if c.startswith('rel_err_width_') and c != 'rel_err_width_total']
base['max_rel_err_decay'] = base[err_cols].max(axis=1, skipna=True)
base['delta_gaga_vs_total'] = (base['rel_err_width_gaga'] - base['rel_err_width_total']).abs()
base['max_rel_err_any'] = base[['max_rel_err_decay','rel_err_width_total']].max(axis=1, skipna=True)

cols_show = [
    'point_id','m_phi','lambda_1','lambda_2','m_12_sq','tan_beta','lambda_6','lambda_7','m_A','m_Hp','sin_ba',
    'rel_err_width_gg','rel_err_width_gaga','rel_err_width_total','max_rel_err_decay','delta_gaga_vs_total','max_rel_err_any'
]

# 1) Top por error total
print('TOP 20 por rel_err_width_total')
display(base.sort_values('rel_err_width_total', ascending=False)[cols_show].head(20))

# 2) Top por diferencia gaga vs total (lo más distinto)
print('TOP 20 por |rel_err_width_gaga - rel_err_width_total|')
display(base.sort_values('delta_gaga_vs_total', ascending=False)[cols_show].head(20))

# 3) Top global (canal de decay o total)
print('TOP 20 por max_rel_err_any')
display(base.sort_values('max_rel_err_any', ascending=False)[cols_show].head(20))
```
