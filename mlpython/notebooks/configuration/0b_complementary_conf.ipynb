```python
import datetime
import math

# Configuración original (espacio de parámetros) como diccionario
previous_parameter_space = {
    "lambda6":    {"min": -6,    "max": 6,      "step": 0.5},
    "lambda7":    {"min": -6,    "max": 6,      "step": 0.5},
    "m12_squared":{"min": -6,    "max": 6,      "step": 0.05},
    "alpha":      {"min": -0.02, "max": 0.01,   "step": 0.05},
    "beta":       {"min": 1.5407960, "max": 1.6007967, "step": 0.01510001},
    "mphi":       {"min": 125,   "max": 400,    "step": 30.314},
    "mA":         {"min": 320.0, "max": 500.0,  "step": 60.0}
}

def calculate_total_points(param_space):
    """Calcula el total de puntos de la grilla para un espacio de parámetros.
       Para cada parámetro se utiliza: floor((max-min)/step)+1, y se multiplica entre ellos.
    """
    total = 1
    for key, spec in param_space.items():
        n = math.floor((spec["max"] - spec["min"]) / spec["step"]) + 1
        total *= n
    return total

def create_complementary_config(original_config):
    """Genera un diccionario con la configuración complementaria desplazada en step/2."""
    comp_config = {}
    for key, spec in original_config.items():
        step = spec["step"]
        comp_config[key] = {
            "min": spec["min"] + step/2.0,
            "max": spec["max"] - step/2.0,
            "step": step
        }
    return comp_config

def save_conf_file(config, filename):
    """Guarda la configuración en un archivo .conf con el formato:
       param_min <value>
       param_max <value>
       step_param <value>
    """
    with open(filename, "w") as f:
        for key in sorted(config.keys()):
            spec = config[key]
            f.write(f"{key}_min {spec['min']}\n")
            f.write(f"{key}_max {spec['max']}\n")
            f.write(f"step_{key} {spec['step']}\n\n")
    print("Archivo de configuración complementaria guardado en:", filename)

def update_log(original_config, comp_config, log_filename):
    """Actualiza un archivo de log (append) con la fecha/hora, la configuración original y la complementaria,
       y el número total de puntos de ambas grillas.
    """
    orig_points = calculate_total_points(original_config)
    comp_points = calculate_total_points(comp_config)
    
    current_time = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    log_lines = []
    log_lines.append("Fecha y hora: " + current_time)
    log_lines.append("Configuración de la grilla original:")
    for key in sorted(original_config.keys()):
        spec = original_config[key]
        log_lines.append(f"  {key}: min={spec['min']}, max={spec['max']}, step={spec['step']}")
    log_lines.append("Configuración de la grilla complementaria:")
    for key in sorted(comp_config.keys()):
        spec = comp_config[key]
        log_lines.append(f"  {key}: min={spec['min']}, max={spec['max']}, step={spec['step']}")
    log_lines.append("Número de puntos en la grilla original: " + str(orig_points))
    log_lines.append("Número de puntos en la grilla complementaria: " + str(comp_points))
    log_lines.append("-" * 40)
    log_entry = "\n".join(log_lines) + "\n"
    
    with open(log_filename, "a") as log_file:
        log_file.write(log_entry)
    print("Registro de log actualizado en:", log_filename)


```

```python
# Generar configuración complementaria a partir de la config previa
complementary_config = create_complementary_config(previous_parameter_space)

# Guardar la nueva configuración en un archivo .conf (por ejemplo, 'complementary_config.conf')
conf_filename = "complementary_config.conf"
save_conf_file(complementary_config, conf_filename)

# Actualizar el archivo de log (por ejemplo, 'grid_log.txt')
log_filename = "grid_log.txt"
update_log(previous_parameter_space, complementary_config, log_filename)
```
